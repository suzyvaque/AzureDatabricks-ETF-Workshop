# 1. Parsing

```python
import fitz

# Test
def extract_first_page_text(pdf_path: str) -> str:
    doc = fitz.open(pdf_path)
    try:
        if doc.page_count == 0:
            return ""
        text = doc[0].get_text("text")
        return text.strip()
    finally:
        doc.close()

text = extract_first_page_text(sample_pdf)
print(text[:3000])
```

```python
import fitz
from typing import List, Dict

def parse_pdf_to_pages(pdf_path: str) -> List[Dict]:
    """
    한 PDF를 페이지 단위로 파싱해서 list[dict] 반환
    """
    doc = fitz.open(pdf_path)
    rows = []
    try:
        for page_num in range(doc.page_count):
            page = doc[page_num]
            text = page.get_text("text") or ""
            rows.append({
                "source_file": os.path.basename(pdf_path),
                "source_path": pdf_path,
                "page_num": page_num + 1,
                "text": text.strip()
            })
        return rows
    finally:
        doc.close()
```

```python
import pandas as pd

all_rows = []
for pdf_path in pdf_files:
    try:
        parsed = parse_pdf_to_pages(pdf_path)
        all_rows.extend(parsed)
        print(f"OK   {os.path.basename(pdf_path)} -> {len(parsed)} pages")
    except Exception as e:
        print(f"FAIL {os.path.basename(pdf_path)} -> {e}")

pdf_pages_pd = pd.DataFrame(all_rows)
print(pdf_pages_pd.shape)
display(pdf_pages_pd.head(20))
```

```python
pdf_pages_sdf = spark.createDataFrame(pdf_pages_pd)
pdf_pages_sdf.printSchema()
```

```python
# Check PDF Quality (empty pages)

from pyspark.sql import functions as F

quality_df = (
    pdf_pages_sdf
    .withColumn("char_len", F.length(F.col("text")))
    .groupBy("source_file")
    .agg(
        F.count("*").alias("page_cnt"),
        F.sum(F.when(F.col("char_len") == 0, 1).otherwise(0)).alias("empty_page_cnt"),
        F.avg("char_len").alias("avg_char_len")
    )
    .orderBy("source_file")
)

display(quality_df)
```

# 2. Chunking

```python
from pyspark.sql import functions as F

catalog = "cat_adb_workshop"
schema = "sch_user0_unstructured"
table_name = f"{catalog}.{schema}.pdf_text"

df = spark.table(table_name)
display(df)
df.printSchema()
```

```python
# page-level text -> smaller chunks

import re
from typing import List, Dict

def normalize_text(text: str) -> str:
    if text is None:
        return ""
    text = text.replace("\xa0", " ")
    text = re.sub(r"\r\n?", "\n", text)
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

def split_by_paragraph(text: str) -> List[str]:
    text = normalize_text(text)
    if not text:
        return []
    parts = [p.strip() for p in re.split(r"\n\s*\n", text) if p.strip()]
    return parts

def force_split_long_text(text: str, max_chars: int = 1000, overlap: int = 150) -> List[str]:
    """
    문단 하나가 너무 길 때 강제로 잘라주는 함수
    """
    text = normalize_text(text)
    if len(text) <= max_chars:
        return [text]

    chunks = []
    start = 0
    step = max_chars - overlap
    if step <= 0:
        raise ValueError("max_chars는 overlap보다 커야 합니다.")

    while start < len(text):
        end = min(start + max_chars, len(text))
        chunk = text[start:end].strip()

        # 가능하면 문장 끝/공백 기준으로 살짝 앞에서 끊기
        if end < len(text):
            candidates = [
                chunk.rfind(". "),
                chunk.rfind("! "),
                chunk.rfind("? "),
                chunk.rfind("\n"),
                chunk.rfind(" ")
            ]
            cut = max(candidates)
            if cut > int(max_chars * 0.6):
                chunk = chunk[:cut + 1].strip()
                end = start + len(chunk)

        if chunk:
            chunks.append(chunk)

        if end >= len(text):
            break
        start = max(end - overlap, start + 1)

    return chunks

def chunk_page_text(
    text: str,
    target_chars: int = 800,
    max_chars: int = 1000,
    overlap: int = 150
) -> List[str]:
    """
    우선 문단 단위로 묶고
    너무 길면 강제 분할
    """
    paragraphs = split_by_paragraph(text)
    if not paragraphs:
        return []

    chunks = []
    current = ""

    for para in paragraphs:
        candidate = para if not current else current + "\n\n" + para

        if len(candidate) <= target_chars:
            current = candidate
            continue

        if current:
            chunks.append(current.strip())
            current = ""

        if len(para) <= max_chars:
            current = para
        else:
            long_parts = force_split_long_text(
                para,
                max_chars=max_chars,
                overlap=overlap
            )
            chunks.extend([p.strip() for p in long_parts if p.strip()])

    if current:
        chunks.append(current.strip())

    # 마지막으로 너무 짧은 청크는 앞 청크에 병합
    merged = []
    for ch in chunks:
        if merged and len(ch) < 120:
            merged[-1] = merged[-1] + "\n\n" + ch
        else:
            merged.append(ch)

    return merged
```

```python
# pdf_pages_pd 기준으로 chunk dataframe 생성

chunk_rows = []

for row in pdf_pages_pd.to_dict(orient="records"):
    page_text = row.get("text", "")
    chunks = chunk_page_text(
        page_text,
        target_chars=800,   # 원하는 chunk 평균 크기
        max_chars=1000,     # 최대 chunk 크기
        overlap=150         # 약간의 문맥 중첩
    )

    for idx, chunk_text in enumerate(chunks, start=1):
        chunk_rows.append({
            "source_file": row["source_file"],
            "source_path": row["source_path"],
            "page_num": row["page_num"],
            "chunk_seq": idx,
            "chunk_text": chunk_text,
            "chunk_char_len": len(chunk_text)
        })

pdf_chunks_pd = pd.DataFrame(chunk_rows)

print(pdf_chunks_pd.shape)
display(pdf_chunks_pd.head(30))
```

```python
# chunk 품질 확인
pdf_chunks_pd["chunk_char_len"].describe()
```

```python
# Spark DataFrame 변환 + chunk_id 생성 (PK 개념)

from pyspark.sql import functions as F

pdf_chunks_sdf = spark.createDataFrame(pdf_chunks_pd)

pdf_chunks_sdf = (
    pdf_chunks_sdf
    .withColumn(
        "chunk_id",
        F.md5(
            F.concat_ws(
                "||",
                F.coalesce(F.col("source_path"), F.lit("")),
                F.coalesce(F.col("page_num").cast("string"), F.lit("")),
                F.coalesce(F.col("chunk_seq").cast("string"), F.lit("")),
                F.coalesce(F.col("chunk_text"), F.lit(""))
            )
        )
    )
    .select(
        "chunk_id",
        "source_file",
        "source_path",
        "page_num",
        "chunk_seq",
        "chunk_text",
        "chunk_char_len"
    )
)

pdf_chunks_sdf.printSchema()
display(pdf_chunks_sdf.orderBy("source_file", "page_num", "chunk_seq"))
```

```python
# 저장

catalog = "cat_adb_workshop"
schema = "sch_user0_unstructured"
chunk_table_name = f"{catalog}.{schema}.pdf_chunks"

(
    pdf_chunks_sdf
    .withColumn("chunked_at", F.current_timestamp())
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(chunk_table_name)
)

print(f"saved to {chunk_table_name}")
display(spark.table(chunk_table_name))
```
